```python
# 📘 visualize_iou.ipynb

# --- 1️⃣ Imports ---
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# --- 2️⃣ Load the CSV ---
csv_path = "results/segmentation_metrics.csv"
df = pd.read_csv(csv_path)
print(df.dtypes)
print(df.head())
```

```python
# --- 3️⃣ Plot with Seaborn ---
sns.set(style="whitegrid", font_scale=1.2)

plt.figure(figsize=(10,6))
sns.lineplot(data=df, x="epoch", y="train_iou", label="Train IoU", marker="o", errorbar=None)
sns.lineplot(data=df, x="epoch", y="val_iou", label="Validation IoU", marker="o", errorbar=None)

# Add test IoU as a horizontal line if present
if "test_iou" in df.columns:
    plt.axhline(df["test_iou"].iloc[-1], color="red", linestyle="--", label=f"Test IoU = {df['test_iou'].iloc[-1]:.4f}")

plt.title("Segmentation IoU over Epochs")
plt.xlabel("Epoch")
plt.ylabel("IoU")
plt.legend()
plt.tight_layout()
plt.show()
```
